```python
print("Welcome to the SpeDiTi Calculator.")

print("\nPlease choose what to calculate:")
print("1 - Speed")
print("2 - Distance (meters)")
print("3 - Time (seconds)")

choice = input("Enter your choice: ")

if choice == "1":
    distance = float(input("Enter distance (m): "))
    time = float(input("Enter time (s): "))

    if time == 0:
        print("Error: Time cannot be zero.")
    else:
        speed = distance / time
        print("Speed =", speed, "m/s")

elif choice == "2":
    speed = float(input("Enter speed (m/s): "))
    time = float(input("Enter time (s): "))

    distance = speed * time
    print("Distance =", distance, "meters")

elif choice == "3":
    speed = float(input("Enter speed (m/s): "))
    distance = float(input("Enter distance (m): "))

    if speed == 0:
        print("Error: Speed cannot be zero.")
    else:
        time = distance / speed
        print("Time =", time, "seconds")

else:
    print("Invalid choice >:( Select a valid choice!")

print("\nThank you for using the SpeDiTi Calculator!")
```
